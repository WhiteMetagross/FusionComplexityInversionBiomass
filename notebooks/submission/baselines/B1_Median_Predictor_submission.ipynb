**Competition:** [CSIRO Image2Biomass](https://www.kaggle.com/competitions/csiro-biomass)

Predicts the training-set median for each target variable.

```python
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd

df_train = pd.read_csv('/kaggle/input/competitions/csiro-biomass/train.csv')
df_test  = pd.read_csv('/kaggle/input/competitions/csiro-biomass/test.csv')

print(f'Train: {df_train.shape}, Test: {df_test.shape}')
```

```python
# Compute median of each target from training data.
median_target_by_name = (
    df_train
    .groupby('target_name')['target']
    .median()
    .reset_index()
)
print('Median Target by Target Name:')
print(median_target_by_name.to_string(index=False))

# Build submission by merging median onto test rows.
df_submission = (
    df_test
    .merge(median_target_by_name, on='target_name', how='left')
    [['sample_id', 'target']]
)
df_submission['target'] = df_submission['target'].fillna(0.0)

print(f'\nSubmission: {df_submission.shape}')
print(df_submission.head(10).to_string(index=False))

df_submission.to_csv('submission.csv', index=False)
print('\nSaved submission.csv')
```
